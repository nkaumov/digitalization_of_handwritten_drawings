# 01 Dataset Registry Audit

This notebook checks the dataset registry and verifies that declared dataset roots and manifests are aligned with the project structure.


```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().resolve().parent if Path.cwd().name == 'notebooks' else Path.cwd().resolve()
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

print(f'PROJECT_ROOT: {PROJECT_ROOT}')
```

```python
import pandas as pd

from src.data import DatasetRegistry
```

```python
registry = DatasetRegistry(PROJECT_ROOT)
registry_df = pd.DataFrame(registry.to_rows())
registry_df
```

```python
audit_rows = []

for row in registry.to_rows():
    root_dir = Path(row['root_dir'])
    manifest_path = Path(row['manifest_path'])

    audit_rows.append({
        'dataset_name': row['dataset_name'],
        'root_exists': root_dir.exists(),
        'manifest_exists': manifest_path.exists(),
        'root_dir': str(root_dir),
        'manifest_path': str(manifest_path),
    })

audit_df = pd.DataFrame(audit_rows)
audit_df
```
