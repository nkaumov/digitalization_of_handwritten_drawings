# 00 Project Bootstrap

This notebook validates the project foundation:
- configuration loading;
- path resolution;
- experiment run directory creation;
- basic artifact directory creation.


```python
from pathlib import Path
import json
import sys

PROJECT_ROOT = Path.cwd().resolve().parent if Path.cwd().name == 'notebooks' else Path.cwd().resolve()
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

print(f'PROJECT_ROOT: {PROJECT_ROOT}')
```

```python
from src.settings import ConfigLoader, PathManager
```

```python
project_config = ConfigLoader.load_project_configs(PROJECT_ROOT)
path_manager = PathManager(PROJECT_ROOT, project_config)
path_manager.ensure_directories()

print('Loaded project configuration successfully.')
print(json.dumps(project_config, indent=2))
```

```python
run_context = path_manager.create_run_context(
    stage=project_config['runtime']['default_experiment_stage'],
    experiment_name=project_config['runtime']['default_experiment_name'],
    runtime_config=project_config,
)

print(f'RUN_ID: {run_context.run_id}')
print(f'RUN_DIR: {run_context.run_dir}')
for artifact_name, artifact_path in run_context.artifact_dirs.items():
    print(f'{artifact_name}: {artifact_path}')
```

```python
config_snapshot_path = run_context.artifact_dirs['config'] / 'project_config_snapshot.json'
config_snapshot_path.write_text(
    json.dumps(project_config, indent=2),
    encoding='utf-8',
)

print(f'Config snapshot saved to: {config_snapshot_path}')
```
